```python
# ============================================
# 1. IMPORT LIBRARIES
# ============================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris


# ============================================
# 2. LOAD DATASET
# ============================================
data = load_iris()

df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target

print("First 5 Rows of Dataset:\n")
print(df.head())


# ============================================
# 3. BASIC INFORMATION
# ============================================
print("\nDataset Info:\n")
print(df.info())

print("\nDataset Shape:", df.shape)


# ============================================
# 4. CHECK MISSING VALUES
# ============================================
print("\nMissing Values:\n")
print(df.isnull().sum())


# ============================================
# 5. DESCRIPTIVE STATISTICS
# ============================================
print("\nDescriptive Statistics:\n")
print(df.describe())


# ============================================
# 6. TARGET VARIABLE ANALYSIS
# ============================================
print("\nTarget Distribution:\n")
print(df['target'].value_counts())


# ============================================
# 7. CORRELATION MATRIX
# ============================================
corr_matrix = df.corr()

print("\nCorrelation Matrix:\n")
print(corr_matrix)


# ============================================
# 8. VISUALIZE CORRELATION MATRIX
# ============================================
plt.figure()

plt.imshow(corr_matrix)
plt.colorbar()

plt.xticks(range(len(corr_matrix.columns)), corr_matrix.columns, rotation=90)
plt.yticks(range(len(corr_matrix.columns)), corr_matrix.columns)

plt.title("Correlation Matrix")
plt.show()


# ============================================
# 9. CORRELATION WITH TARGET
# ============================================
print("\nCorrelation with Target:\n")
target_corr = corr_matrix['target'].sort_values(ascending=False)
print(target_corr)


# ============================================
# 10. IDENTIFY IMPORTANT FEATURES
# ============================================
print("\nImportant Features (based on correlation magnitude):")

for feature, value in target_corr.items():
    if feature != 'target':
        print(f"{feature} --> Correlation: {value:.3f}")


# ============================================
# 11. INTERPRETATION LOGIC
# ============================================
print("\nInterpretation:")
print("- High positive correlation → strong direct relationship")
print("- High negative correlation → inverse relationship")
print("- Near zero → weak or no linear relationship")
```
