```python
text = """Artificial Intelligence (AI) is the simulation of human intelligence in machines that are programmed to think and perform tasks such as learning, reasoning, and problem-solving.

Machine Learning is a subset of Artificial Intelligence that allows systems to automatically learn and improve from experience without being explicitly programmed.

Deep Learning is a specialized subset of Machine Learning that uses artificial neural networks to model and solve complex problems.

Natural Language Processing (NLP) enables machines to understand and generate human language.

Internet of Things (IoT) refers to interconnected devices that communicate over the internet.

Retrieval-Augmented Generation (RAG) enhances language model responses by retrieving relevant information from documents before generating answers.

Embeddings are numerical representations of text that capture semantic meaning.

Cosine similarity measures how similar two text vectors are.
"""

with open("data.txt", "w") as f:
    f.write(text)
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Load model
model = SentenceTransformer('all-MiniLM-L6-v2')

# Load document
with open("data.txt", "r") as file:
    text = file.read()

# Split into chunks
chunks = text.split("\n")

# Create embeddings
chunk_embeddings = model.encode(chunks)

print("✅ RAG System Ready!\n")

# Input query
query = input("Enter your question: ")

# Query embedding
query_embedding = model.encode([query])

# Similarity calculation
similarities = cosine_similarity(query_embedding, chunk_embeddings)

# Best match
best_index = np.argmax(similarities)
best_chunk = chunks[best_index]

# Professional Output
print("\n" + "="*50)
print("        RETRIEVAL-AUGMENTED GENERATION RESULT")
print("="*50)

print("\nQuery:")
print(query)

print("\nRetrieved Context:")
print(best_chunk)

print("\nGenerated Response:")
print(f"Based on the retrieved information, it can be concluded that {best_chunk}")

print("\nStatus: SUCCESS")
print("="*50)
```
