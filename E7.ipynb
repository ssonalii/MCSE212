```python
# ============================================
# EXPERIMENT: AI-ASSISTED CODE GENERATION
# Sorting using Bubble Sort (Bug → Fix → Optimize)
# ============================================

# ============================================
# 1. INPUT DATA
# ============================================
data = [64, 34, 25, 12, 22, 11, 90]

print("============================================")
print("ORIGINAL INPUT")
print("============================================")
print("Original List:", data)


# ============================================
# 2. AI-GENERATED CODE (BUGGY VERSION)
# ============================================
def bubble_sort_buggy(arr):
    n = len(arr)
    for i in range(n):
        for j in range(n):   # BUG HERE
            if arr[j] > arr[j+1]:   # Index error
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr


print("\n============================================")
print("BUGGY VERSION OUTPUT")
print("============================================")

try:
    result_buggy = bubble_sort_buggy(data.copy())
    print("Sorted List:", result_buggy)
except Exception as e:
    print("Error occurred:", e)


# ============================================
# 3. DEBUGGED VERSION
# ============================================
def bubble_sort_fixed(arr):
    n = len(arr)
    for i in range(n):
        for j in range(n - 1):   # FIXED
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr


print("\n============================================")
print("DEBUGGED VERSION OUTPUT")
print("============================================")

result_fixed = bubble_sort_fixed(data.copy())
print("Sorted List:", result_fixed)


# ============================================
# 4. OPTIMIZED VERSION
# ============================================
def bubble_sort_optimized(arr):
    n = len(arr)
    for i in range(n):
        swapped = False

        for j in range(0, n - i - 1):   # Reduced comparisons
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True

        if not swapped:   # Early stopping
            break

    return arr


print("\n============================================")
print("OPTIMIZED VERSION OUTPUT")
print("============================================")

result_optimized = bubble_sort_optimized(data.copy())
print("Optimized Sorted List:", result_optimized)


# ============================================
# 5. PERFORMANCE COMPARISON
# ============================================
import time

print("\n============================================")
print("PERFORMANCE COMPARISON")
print("============================================")

start = time.time()
bubble_sort_fixed(data.copy())
end = time.time()
print("Time (Debugged):", round((end - start)*1000000, 2), "microseconds")

start = time.time()
bubble_sort_optimized(data.copy())
end = time.time()
print("Time (Optimized):", round((end - start)*1000000, 2), "microseconds")


# ============================================
# 6. FINAL VERIFICATION
# ============================================
print("\n============================================")
print("FINAL VERIFICATION")
print("============================================")

if result_fixed == result_optimized:
    print("Both methods produce the SAME correct result.")
else:
    print("Mismatch detected!")
```
