```python
# ============================================
# EXPERIMENT 5: EFFECT OF PROMPT VARIATION
# (NO API VERSION - SELF CONTAINED SIMULATION)
# ============================================

import pandas as pd

# ============================================
# 1. INPUT TEXT (COMMON FOR ALL PROMPTS)
# ============================================
input_text = """
Machine learning is a subset of artificial intelligence that focuses on enabling systems
to learn from data. It involves algorithms that improve automatically through experience.
Applications include recommendation systems, image recognition, and natural language processing.
"""

# ============================================
# 2. DEFINE PROMPTS
# ============================================
prompts = {
    "Basic Prompt": "Summarize the text without structure.",

    "Structured Prompt": """
Summarize the text in 3 bullet points:
- Keep it concise
- Use simple language
""",

    "Constrained Prompt": """
Summarize in max 30 words including 'machine learning'.
Must be single paragraph.
""",

    "Explanation Prompt": """
Explain for beginner using analogy under 50 words.
"""
}

# ============================================
# 3. SIMULATED MODEL OUTPUTS
# (Realistic ChatGPT-like responses)
# ============================================

responses = {
    "Basic Prompt":
        "Machine learning is a part of AI that allows systems to learn from data and improve over time. It is used in many applications like recommendation systems and image recognition.",

    "Structured Prompt":
        "- Machine learning is a branch of AI.\n"
        "- It enables systems to learn from data.\n"
        "- It is used in applications like NLP and image recognition.",

    "Constrained Prompt":
        "Machine learning is a subset of AI that enables systems to learn from data for tasks like recommendation and image recognition.",

    "Explanation Prompt":
        "Machine learning is like teaching a child using examples. Instead of giving rules, we show data so the system learns patterns and makes predictions."
}

# ============================================
# 4. DISPLAY OUTPUTS
# ============================================
print("\n========== GENERATED OUTPUTS ==========\n")

for k, v in responses.items():
    print(f"\n----- {k} -----\n")
    print(v)

# ============================================
# 5. QUANTITATIVE ANALYSIS
# ============================================
analysis = []

for k, v in responses.items():
    words = v.split()

    analysis.append({
        "Prompt Type": k,
        "Word Count": len(words),
        "Character Count": len(v),
        "Contains ML Term": "machine learning" in v.lower(),
        "Structure Level": "High" if "-" in v else "Low"
    })

df = pd.DataFrame(analysis)

print("\n========== QUANTITATIVE ANALYSIS ==========\n")
print(df)

# ============================================
# 6. PROMPT EFFECT ANALYSIS (CORE RESULT)
# ============================================
print("\n========== EFFECT OF PROMPT VARIATION ==========\n")

print("1. Basic Prompt:")
print("   → Produces natural but unstructured output.\n")

print("2. Structured Prompt:")
print("   → Improves readability using bullet format.\n")

print("3. Constrained Prompt:")
print("   → Ensures strict word limit and keyword inclusion.\n")

print("4. Explanation Prompt:")
print("   → Converts technical concept into analogy-based understanding.\n")

print("Conclusion:")
print("Prompt clarity and constraints directly control output structure, length, and interpretability.")

# ============================================
# 7. FINAL COMPARISON SUMMARY
# ============================================
print("\n========== FINAL SUMMARY TABLE ==========\n")
print(df)

# ============================================
# 8. SAVE RESULTS
# ============================================
df.to_csv("prompt_variation_results.csv", index=False)

print("\nResults saved as prompt_variation_results.csv")
```
